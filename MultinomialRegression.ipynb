```python
#imports 
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.model_selection import cross_val_score
from sklearn.preprocessing import PowerTransformer
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
```

```python
df=pd.read_csv("drug.csv")
print(df.index)
print("All columns are")
ai=0
for i in df:
    print(ai,i)
    ai+=1
print(df.columns)
```

```python
# testing without encoding
tdf = df.iloc[:,0:5]
ttdf = df.iloc[:,5:6]
print(tdf)
print(ttdf)
```

```python
le = LabelEncoder()
al = df
tdf['sex'] = le.fit_transform(tdf['sex'])
tdf['bp'] = le.fit_transform(tdf['bp'])
tdf['cholesterol'] = le.fit_transform(tdf['cholesterol'])
ttdf['drug'] = le.fit_transform(ttdf['drug'])
ttdf = ttdf['drug']
print(tdf)
print(ttdf)
```

```python
tX_train, tX_test, ty_train, ty_test = train_test_split( tdf, ttdf,test_size=0.2)

# model = LogisticRegression(multi_class='multinomial',solver='lbfgs',max_iter=1000)
# We dont need to specify multinomial it auto detects

tmodel = LogisticRegression(max_iter=2000)

tmodel.fit(tX_train, ty_train)


ty_pred = tmodel.predict(tX_test)


print("Accuracy:", accuracy_score(ty_test, ty_pred))

print("\nClassification Report:")
print(classification_report(ty_test, ty_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(ty_test, ty_pred))
```

```python
tscores = cross_val_score(tmodel, tdf, ttdf, cv=10, scoring="accuracy")

print('Actual cross-val results')
print(f"Scores for each fold: {tscores}")
print(f"Mean Accuracy: {tscores.mean():.2f}")
print(f"Standard Deviation: {tscores.std():.2f}")
```

```python
# lets check for one hot encoding and scaling
X = df[['age', 'sex', 'bp', 'cholesterol', 'Na_to_K']]
y = df['drug']
print(X)
print(y)
```

```python
le = LabelEncoder()
y = le.fit_transform(y)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2,)


preprocessor = ColumnTransformer([
    ('yeo_johnson', PowerTransformer(method='yeo-johnson', standardize=True), ['age', 'Na_to_K']),
    ('onehot', OneHotEncoder(drop='first', handle_unknown='ignore', sparse_output=False), ['sex', 'bp', 'cholesterol'])
], remainder='passthrough')


pipeline = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', LogisticRegression(solver='lbfgs'))
])

pipeline.fit(X_train, y_train)
```

```python
X_train_transformed = pipeline.named_steps['preprocessor'].transform(X_train)

feature_names = pipeline.named_steps['preprocessor'].get_feature_names_out()

transformed_df = pd.DataFrame(
    X_train_transformed,
    columns=feature_names
)

print(transformed_df.head())

print()

pt = pipeline.named_steps['preprocessor'].named_transformers_['yeo_johnson']

print(pt.lambdas_)
```

```python
y_pred = pipeline.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
```

```python
from sklearn.metrics import classification_report, confusion_matrix

print("Accuracy:", accuracy_score(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
# Checking through pipeline
scores = cross_val_score(pipeline ,X ,y ,cv=10,scoring='accuracy')

print("Cross Validation Scores:", scores)
print("Mean Accuracy:", scores.mean())
print("Standard Deviation:", scores.std())
```
